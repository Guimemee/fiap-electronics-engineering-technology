> Integrar os módulos de sensoriamento, processador ARM Cortex-M e rádio transmissor LoRaWAN em regime de ultra baixo consumo (Deep Sleep) para autonomia de 5 anos em bateria.

### Formulação Matemática Formal
$$I_{médio} = \frac{I_{ativo} \cdot t_{ativo} + I_{sleep} \cdot t_{sleep}}{t_{ciclo}}, \quad \text{Autonomia} = \frac{C_{bateria}}{I_{médio}}$$


### Passo 1: Perfil de Consumo Ativo
No ciclo de transmissão (1 segundo a cada 10 minutos): corrente ativa I_{ativo} = 35 \text{ mA}.


### Passo 2: Perfil de Consumo em Modo Hibernação (Deep Sleep)
Durante o intervalo inativo de 599 segundos: corrente residual I_{sleep} = 4,2 \, \mu\text{A}.


### Passo 3: Corrente Média Ponderada
I_{médio} = \frac{35 \text{ mA} \times 1\text{s} + 0,0042 \text{ mA} \times 599\text{s}}{600\text{s}} = \frac{35 + 2,516}{600} \approx 62,5 \, \mu\text{A}.


### Passo 4: Cálculo da Vida Útil da Bateria
Com célula LiSOCl_2 de 3.6 V / 3000 mAh e desclassificação de 80\%: Vida = \frac{2400 \text{ mAh}}{0,0625 \text{ mA}} = 38.400 \text{ horas} \approx 4,38 \text{ anos}.


### Passo 5: Conclusão do Desafio Natura
Sistema 100% autônomo e hermético operando na selva sem manutenção recarregável.


```python
import numpy as np
import matplotlib.pyplot as plt

intervalos_min = np.array([1, 5, 10, 15, 30, 60])
i_medio_ua = (35000 * 1 + 4.2 * (intervalos_min * 60 - 1)) / (intervalos_min * 60)
vida_anos = (2400 * 1000 / i_medio_ua) / (24 * 365)
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#211208')
ax.set_facecolor('#211208')
ax.plot(intervalos_min, vida_anos, 'o-', color='#f97316', lw=2.5, label='Autonomia da Bateria (Anos)')
ax.scatter([10], [4.38], color='#10b981', s=70, zorder=5, label='Operação Natura (10 min -> 4.4 anos)')
ax.set_title('Autonomia do Nó LoRaWAN vs Intervalo de Transmissão', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Intervalo entre Transmissões (minutos)', color='#94a3b8')
ax.set_ylabel('Vida Útil Estimada (Anos)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#4d2b14', alpha=0.5)
ax.legend(facecolor='#331c0c', edgecolor='#f97316', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
