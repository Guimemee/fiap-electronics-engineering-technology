> **Enunciado Original:**
> Implementar a temporização de bit do controlador CAN (Bit Timing) para barramento a 500 kbps, configurando os segmentos Sync_Seg, Prop_Seg, Phase_Seg1 e Phase_Seg2.

### Formulação Matemática Formal
$$t_{bit} = TQ \times (\text{Sync} + \text{Prop} + \text{Phase1} + \text{Phase2}), \quad \text{Sample Point} \approx 87,5\%$$


### Passo 1: Tempo Total de 1 Bit
Para taxa de 500 kbps: t_{bit} = \frac{1}{500.000} = 2,0 \mu\text{s}.


### Passo 2: Definição do Quantum de Tempo (TQ)
Com clock de periférico f_{APB1} = 42 \text{ MHz} e prescaler BRP = 6: TQ = \frac{6}{42 \text{ MHz}} = 142,85 \text{ ns}.


### Passo 3: Distribuição dos Quanta de Tempo
Total de 14 TQ: Sync_Seg = 1 TQ; Prop_Seg = 6 TQ; Phase_Seg1 = 5 TQ; Phase_Seg2 = 2 TQ.


### Passo 4: Verificação do Ponto de Amostragem (Sample Point)
\text{Sample Point} = \frac{1 + 6 + 5}{14} = \frac{12}{14} \approx 85,7\%, em perfeita conformidade com as normas CiA.


### Passo 5: Conclusão da Rede
Comunicação diferencial robusta com transceptores MCP2551 garantindo alcance de até 100 m na instalação industrial.


```python
import numpy as np
import matplotlib.pyplot as plt

segmentos = ['Sync_Seg (1 TQ)', 'Prop_Seg (6 TQ)', 'Phase_Seg1 (5 TQ)', 'Phase_Seg2 (2 TQ)']
quanta = [1, 6, 5, 2]
cores = ['#38bdf8', '#f97316', '#eab308', '#ef4444']
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#211208')
ax.set_facecolor('#211208')
bars = ax.barh(segmentos, quanta, color=cores, height=0.45)
ax.set_title('Composição do Bit Timing no Protocolo CAN (14 TQ)', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Quanta de Tempo (TQ)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(axis='x', color='#4d2b14', alpha=0.5)
for bar in bars:
    wval = bar.get_width()
    ax.text(wval + 0.15, bar.get_y() + bar.get_height()/2.0, f'{wval} TQ', ha='left', va='center', color='#ffffff', fontweight='bold')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
