**Título:** Barramentos Seriais de Alta Velocidade: Comparativo SPI vs I2C

> **Enunciado Original:**
> Dimensionar os resistores de pull-up do barramento I2C em modo Fast-Mode (400 kHz) para compensar a capacitância parasita do cabo de telemetria C_bus = 200 pF.

### Formulação Matemática Formal
$$R_{p(mín)} = \frac{V_{cc} - V_{OL}}{I_{OL}} = \frac{3,3 - 0,4}{3 \times 10^{-3}} \approx 967 \, \Omega, \quad R_{p(máx)} = \frac{t_r}{0,8473 \cdot C_b}$$


### Passo 1: Tempo Máximo de Subida (Rise Time) no Fast-Mode
A especificação I2C Fast-Mode (400 kHz) estabelece tempo máximo de subida t_r \le 300 \text{ ns}.


### Passo 2: Resistência Máxima de Pull-Up
R_{p(máx)} = \frac{300 \times 10^{-9}}{0,8473 \times 200 \times 10^{-12}} = \frac{300}{0,16946} \approx 1.770 \Omega.


### Passo 3: Resistência Mínima de Pull-Up
Para garantir que a saída em nível baixo não drene mais que I_{OL} = 3 \text{ mA}: R_{p(mín)} = \frac{3,3 - 0,4}{0,003} \approx 967 \Omega.


### Passo 4: Seleção do Valor Comercial E24
Adotado R_p = 1,5 \text{ k}\Omega, satisfazendo a faixa [967 \Omega, 1770 \Omega].


### Passo 5: Conclusão do Barramento
Sinais SDA e SCL apresentam bordas limpas sem overshoot e taxa de erros de transmissão zero.


```python
import numpy as np
import matplotlib.pyplot as plt

c_bus = np.linspace(50, 400, 50)
rp_max = 300e-9 / (0.8473 * c_bus * 1e-12)
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#211208')
ax.set_facecolor('#211208')
ax.plot(c_bus, rp_max, color='#f97316', lw=2.5, label='Rp Máximo Permitido (Ω)')
ax.axhline(967, color='#ef4444', linestyle='--', label='Rp Mínimo Limite (967 Ω)')
ax.scatter([200], [1770], color='#10b981', s=60, zorder=5, label='Ponto de Projeto (200 pF, 1.77kΩ)')
ax.set_title('Dimensionamento de Pull-Up I2C Fast-Mode', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Capacitância do Barramento Cb (pF)', color='#94a3b8')
ax.set_ylabel('Resistência Pull-Up (Ω)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#4d2b14', alpha=0.5)
ax.legend(facecolor='#331c0c', edgecolor='#f97316', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
