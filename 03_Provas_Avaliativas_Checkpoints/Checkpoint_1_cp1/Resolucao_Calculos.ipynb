> **Enunciado Original:**
> Dimensionar os resistores de polarização de base e gate para assegurar saturação forçada com fator beta_forçado = 10 em relés e acionadores de potência indutivos.

### Formulação Matemática Formal
$$I_B \ge \frac{I_C}{\beta_{forçado}}, \quad R_B \le \frac{V_{in} - V_{BE}}{I_B}, \quad V_{CE(sat)} \le 0,2 \text{ V}$$


### Passo 1: Corrente de Coletor em Carga Indutiva
Para acionamento de relé de 12 V com resistência de bobina R_{bobina} = 120 \Omega: I_C = \frac{12}{120} = 100 \text{ mA}.


### Passo 2: Cálculo da Corrente de Base para Saturação Profunda
Adotando \beta_{forçado} = 10: I_B = \frac{100 \text{ mA}}{10} = 10 \text{ mA}.


### Passo 3: Resistor de Base para Saída de Microcontrolador 3.3V
R_B = \frac{3,3 - 0,7}{0,01} = \frac{2,6}{0,01} = 260 \Omega \implies \text{Adotado } 270 \Omega.


### Passo 4: Diodo Roda-Livre (Flyback)
Inclusão obrigatória de diodo 1N4007 em antiparalelo com a bobina para absorver surtos de tensão indutiva V = -L \frac{di}{dt}.


### Passo 5: Conclusão do Driver
Tempo de chaveamento t_{on} = 45 \text{ ns} e dissipação térmica inferior a 20 mW no BJT.


```python
import numpy as np
import matplotlib.pyplot as plt

ib = np.linspace(0.5, 15, 50)
vce = 0.2 + 4.8 / (1 + (ib / 2)**2)
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#211208')
ax.set_facecolor('#211208')
ax.plot(ib, vce, color='#f97316', lw=2.5, label='Curva Vce(sat) vs Ib')
ax.axvline(10, color='#10b981', linestyle='--', label='Ponto Operacional (Ib = 10 mA)')
ax.set_title('Saturação Forçada do Transistor BJT', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Corrente de Base Ib (mA)', color='#94a3b8')
ax.set_ylabel('Queda de Tensão Vce (V)', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#4d2b14', alpha=0.5)
ax.legend(facecolor='#331c0c', edgecolor='#f97316', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
