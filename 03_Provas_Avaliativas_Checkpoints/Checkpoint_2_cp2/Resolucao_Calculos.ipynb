**Título:** Condicionamento e Aquisição de Sinais com ADC SAR de 12 Bits

> **Enunciado Original:**
> Calcular a resolução de tensão, o erro de quantização e dimensionar o estágio de proteção com diodos zener para um conversor ADC de 12 bits com tensão de referência V_ref = 3.3 V.

### Formulação Matemática Formal
$$LSB = \frac{V_{ref}}{2^N - 1} = \frac{3,3}{4095} \approx 805,86 \, \mu\text{V}, \quad SNR = 6,02 N + 1,76 \text{ dB} \approx 74 \text{ dB}$$


### Passo 1: Quantização e LSB (Least Significant Bit)
Com N = 12 bits e 4096 níveis de quantização, 1 \text{ LSB} = \frac{3,3}{4095} \approx 805,86 \mu\text{V}.


### Passo 2: Erro Máximo de Quantização
E_q = \pm \frac{1}{2} LSB \approx \pm 402,9 \mu\text{V}.


### Passo 3: Relação Sinal-Ruído Teórica (SNR)
SNR = 6,02(12) + 1,76 = 74,0 \text{ dB}.


### Passo 4: Circuito de Proteção e Clamp
Resistor limitador de 1 \text{ k}\Omega e diodo Schottky BAT54S limitando tensões de entrada estritamente entre -0,3 V e 3,6 V.


### Passo 5: Conclusão Metrológica
O sistema garante precisão de leitura de sensores de temperatura na floresta com resolução de 0,05 °C.


```python
import numpy as np
import matplotlib.pyplot as plt

tensao_in = np.linspace(0, 3.3, 100)
codigos = np.floor(tensao_in / 3.3 * 4095)
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#211208')
ax.set_facecolor('#211208')
ax.plot(tensao_in, codigos, color='#f97316', lw=2, label='Código Digital ADC (0 a 4095)')
ax.set_title('Curva de Conversão do ADC SAR 12-Bit', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Tensão Analógica de Entrada (V)', color='#94a3b8')
ax.set_ylabel('Código Digital Inteiro', color='#94a3b8')
ax.tick_params(colors='#94a3b8')
ax.grid(True, color='#4d2b14', alpha=0.5)
ax.legend(facecolor='#331c0c', edgecolor='#f97316', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
