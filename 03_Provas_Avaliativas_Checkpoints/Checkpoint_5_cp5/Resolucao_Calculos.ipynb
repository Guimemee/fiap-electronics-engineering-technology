**Título:** Controle de Atuadores via PWM em Ponte H com Proteção Dead-Time

> **Enunciado Original:**
> Dimensionar o tempo morto (Dead-Time) entre os sinais de disparo dos transistores high-side e low-side da Ponte H para prevenir curto-circuito em condução cruzada (Shoot-Through).

### Formulação Matemática Formal
$$t_{dead} \ge t_{off(max)} - t_{on(min)} + t_{prop(delay)} \approx 250 \text{ ns}$$


### Passo 1: Dinâmica de Chaveamento dos MOSFETs
Tempo de desligamento máximo t_{off} = 180 \text{ ns}; tempo de subida mínimo t_{on} = 40 \text{ ns}.


### Passo 2: Atraso de Propagação do Gate Driver
O driver IR2104 introduz skew de propagação de \approx 50 \text{ ns}.


### Passo 3: Margem de Segurança Térmica
Adotar margem de proteção de 50 ns para variações de temperatura até 85 °C.


### Passo 4: Tempo Morto Mínimo Resultante
t_{dead} \ge 180 - 40 + 50 + 50 = 240 \text{ ns} \implies \text{Configurado timer com } 300 \text{ ns}.


### Passo 5: Conclusão de Eficiência Mecânica
Controle bidirecional suave de motor DC de 24 V sem aquecimento residual nos MOSFETs.


```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 1000, 500)
high_side = np.where((t >= 100) & (t <= 450), 1, 0)
low_side = np.where((t >= 550) & (t <= 900), 1, 0)
fig, ax = plt.subplots(figsize=(6, 3.5), facecolor='#211208')
ax.set_facecolor('#211208')
ax.plot(t, high_side, color='#f97316', lw=2.5, label='MOSFET High-Side')
ax.plot(t, low_side, color='#38bdf8', lw=2.5, label='MOSFET Low-Side')
ax.axvspan(450, 550, color='#ef4444', alpha=0.3, label='Dead-Time (100 ns)')
ax.set_title('Proteção por Dead-Time em Ponte H (Shoot-Through Free)', color='#ffffff', fontsize=11, fontweight='bold')
ax.set_xlabel('Tempo (ns)', color='#94a3b8')
ax.set_ylabel('Nível Lógico de Disparo', color='#94a3b8')
ax.set_ylim(-0.2, 1.4)
ax.tick_params(colors='#94a3b8')
ax.legend(facecolor='#331c0c', edgecolor='#f97316', labelcolor='#ffffff')
plt.tight_layout()
plt.savefig('grafico_resolucao.png', dpi=120)
plt.close()
```
